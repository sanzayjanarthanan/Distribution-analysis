```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")
df.to_csv("iris.csv", index=False)
data = df["petal_length"]

print(data.describe())
print("Skewness:", round(data.skew(), 2))

plt.figure(figsize=(8,5))
plt.hist(data, bins=15, color="steelblue", edgecolor="black")
plt.axvline(data.mean(), color="red", linestyle="--", label=f"Mean = {data.mean():.2f}")
plt.axvline(data.median(), color="green", linestyle="--", label=f"Median = {data.median():.2f}")
plt.title("Distribution of Petal Length (Iris Dataset)")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.legend()
plt.savefig("histogram.png", dpi=150, bbox_inches="tight")
plt.show()
```
